```python
import os
import pandas as pd
from bs4 import BeautifulSoup
```

```python
soup = BeautifulSoup(open("../../data/drugbank/Drugbank_Fulldata.xml"),"xml")
```

```python
drugs = soup.findAll("drug")
```

```python
data = []
for idx, drug in enumerate(drugs):
    print(f'{idx}/{len(drugs)}', end='\r')
    drug_id = drug.find("drugbank-id").text
    drug_name = drug.find("name").text
    data.append({
        'id': drug_id,
        'name': drug_name
    })
    
df = pd.DataFrame(data)
df = df.dropna()
df = df.drop_duplicates()
df.to_csv('../../data/drugbank/drugs.csv', index=False)
```

```python
data = []

for idx, drug in enumerate(drugs):
    print(f'{idx}/{len(drugs)}', end='\r')
    drug_id = drug.find("drugbank-id").text
    drug_name = drug.find("name").text
    pathways = drug.find('pathways')
    if pathways:
        for pathway in pathways.findAll('pathway'):
            if pathway:
                smpdb_id = pathway.find('smpdb-id').text
                pathway_name = pathway.find('name').text
                data.append({
                    'DrugBankID': drug_id,
                    'DrugBankName': drug_name,
                    'relation': 'drug_pathway',
                    'SMPDBID': smpdb_id,
                    'PathwayName': pathway_name
                })
                drugs_inner = pathway.find('drugs')
                if drugs_inner:
                    for drug_inner in drugs_inner.findAll('drug'):
                        data.append({
                            'DrugBankID': drug_inner.find("drugbank-id").text,
                            'DrugBankName': drugs_inner.find("name").text,
                            'relation': 'drug_pathway',
                            'SMPDBID': smpdb_id,
                            'PathwayName': pathway_name
                        })
df = pd.DataFrame(data)
print(df.shape)
df = df.dropna()
df = df.drop_duplicates(subset=['DrugBankID', 'SMPDBID'])
df.head()
```
