```python
from unike.module.model import TransE
import yaml
import sys

sys.path.extend(['.', '..'])

from q import plot_tsne
```

```python
# need to change
config_path = '../../config/transe_gpus_20250910.yaml'
checkpoint_path = '../../checkpoints/transe_gpus_20250910-200.pth'
```

```python
with open(config_path, 'r') as f:
	config = yaml.load(f, Loader=yaml.FullLoader)

model = TransE(
	ent_tol = 121649,
	rel_tol = 22,
	dim = config['dim'], 
	p_norm = config['p_norm'], 
	norm_flag = config['norm_flag']
)
model.load_checkpoint(checkpoint_path)
```

```python
embeddings = model.ent_embeddings.weight.data.cpu().numpy()
plot_tsne(embeddings)
```
