```python
from unike.utils import Link
from unike.module.model import RotatE, ComplEx
import pandas as pd
```

```python
ent_tol = 151321
rel_tol = 31
```

```python
rotate_model = RotatE(
	ent_tol = ent_tol,
	rel_tol = rel_tol,
	dim = 1024,
	margin = 6.0,
	epsilon = 2.0,
)
```

```python
rotate_model.load_checkpoint('checkpoints/rotate/all/base/rotate-200.pth')
```

```python
link = Link(
    in_path='data',
    all_file='all2id_base.txt',
    model=rotate_model
)
```

```python
all_drug_ent_id = [link.ent2id[ent_name] for ent_name in link.ent2id.keys() if ent_name.split(':')[-1] == 'drug']
len(all_drug_ent_id)
```

```python
nodes = pd.read_csv('../../../kg/nodes.csv')
```

```python
df = link.link([5153], [1], all_drug_ent_id, device='cuda:3')
df = pd.merge(df, nodes, left_on=['tail'], right_on=['node_index'], how='left').drop(columns=['node_index', 'node_type', 'node_name']).rename(columns={'node_id': 'tail_id', 'node_source': 'tail_source'})
df.to_csv('link_result_rotate_200_kg4rd_base_S1PR1_drug.csv', index=False)
```
