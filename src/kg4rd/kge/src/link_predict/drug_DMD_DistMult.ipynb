```python
from unike.module.model import DistMult
import yaml
import sys

sys.path.extend(['.', '..'])

from q import link, drug_ent_indexs, indication_rel_index, dmd_ent_index, add_id
```

```python
config_path = '../../config/DistMult/DistMult_base_eval_Accel_20251118.yaml'
checkpoint_path = '../../checkpoints/DistMult_base_eval_Accel_20251118/model-400.pth'
```

```python
with open(config_path, 'r') as f:
	config = yaml.load(f, Loader=yaml.FullLoader)

model = DistMult(
	ent_tol = 121649,
	rel_tol = 22,
	dim = config['dim']
)
model.load_checkpoint(checkpoint_path)
```

```python
df = link.link(
    drug_ent_indexs,
    [indication_rel_index],
    [dmd_ent_index],
    model, 'cuda:0'
)
df = add_id(df)
df.to_csv('../../link_result/drug_DMD_DistMult.csv', index=False)
```
