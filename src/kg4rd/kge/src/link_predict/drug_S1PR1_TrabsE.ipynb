```python
from unike.module.model import TransE
import yaml
import sys

sys.path.extend(['.', '..'])

from q import *
from config import ent_tol, rel_tol
```

```python
# need to change
config_path = '../../config/transe_gpus_20250910.yaml'
checkpoint_path = '../../checkpoints/transe_gpus_20250910-200.pth'
```

```python
with open(config_path, 'r') as f:
	config = yaml.load(f, Loader=yaml.FullLoader)

model = TransE(
	ent_tol = ent_tol,
	rel_tol = rel_tol,
	dim = config['dim'], 
	p_norm = config['p_norm'], 
	norm_flag = config['norm_flag']
)
model.load_checkpoint(checkpoint_path)
```

```python
df = link.link(
    drug_ent_indexs,
    [drug_protein_rel_index],
    [s1pr1_ent_index],
    model, 'cuda:0'
)
df = add_id(df)
df.to_csv('../../link_result/drug_s1pr1_transe.csv', index=False)
```
