```python
import pandas as pd
import sys

sys.path.append('.')

from q import nodes
from q import edges, edges_supplement
```

```python
total_node = nodes.shape[0]
print("实体总数量:", total_node)
node_counts = nodes['node_type'].value_counts()
node_percent = (node_counts / total_node * 100).round(3)
node_summary = pd.DataFrame({
    '实体数量': node_counts,
    '占比(%)': node_percent
})
print("不同类型实体数量及占比:")
print(node_summary)
```

```python
edges = pd.concat([
    edges,
    edges_supplement
], ignore_index=True)
```

```python
total_relations = edges.shape[0]
print("关系总数量:", total_relations)
relation_counts = edges['relation'].value_counts()
relation_percent = (relation_counts / total_relations * 100).round(3)
relation_summary = pd.DataFrame({
    '关系数量': relation_counts,
    '占比(%)': relation_percent
})
print("不同类型关系数量及占比:")
print(relation_summary)
```
