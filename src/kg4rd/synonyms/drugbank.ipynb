```python
import os
import pandas as pd
from bs4 import BeautifulSoup
```

```python
soup = BeautifulSoup(open("../../../data/drugbank/Drugbank_Fulldata.xml"),"xml")
```

```python
drugs = soup.findAll("drug")
```

```python
data = []

for idx, drug in enumerate(drugs):
    print(f'{idx}/{len(drugs)}', end='\r')
    id = drug.find("drugbank-id").text
    name = drug.find("name").text

    data.append({
        'id': id,
        'name': name,
        'preferred_name': True
    })
    synonyms = drug.find("synonyms")    
    if synonyms is not None:
        for synonym in synonyms.find_all("synonym"):
            data.append({
                'id': id,
                'name': synonym.text.strip(),
                'preferred_name': ''
            })

df = pd.DataFrame(data)
df.drop_duplicates(inplace=True)
df.to_csv('../../../data_synonyms/drugbank_synonyms.csv', index=False)
```
