```python
import pandas as pd
```

```python
df = pd.read_csv('../../../data_synonyms/mondo_synonyms_concat.csv')
synonyms_count = df.shape[0]
node_count = df['id'].drop_duplicates().count()
print("mondo synonyms count: ", synonyms_count)
print("mondo node count: ", node_count)
print("mondo mean value: ", synonyms_count / node_count)
df.head()
```
