```python
from unike.utils import Link
from unike.module.model import RotatE, ComplEx
import pandas as pd
```

```python
ent_tol = 129375
rel_tol = 30
```

```python
rotate_model = RotatE(
    ent_tol=ent_tol,
    rel_tol=rel_tol,
    dim=1024,
    margin=6.0,
    epsilon=2.0,
)
```

```python
rotate_model.load_checkpoint('checkpoints/rotate/all/rotate-200.pth')
```

```python
link = Link(
    in_path='../../primekg/kg/',
    model=rotate_model
)
```

```python
dmd_ent_id = [32149]
drug_dis_rel_id = [3]
all_drug_ent_id = [link.ent2id[ent_name] for ent_name in link.ent2id.keys() if ent_name.split(':')[-1] == 'drug']
len(all_drug_ent_id)
```

```python
nodes = pd.read_csv('../../primekg/kg/nodes.csv')
```

```python
df = link.link(dmd_ent_id, drug_dis_rel_id, all_drug_ent_id, device='cuda:3')
df = pd.merge(df, nodes, left_on=['tail'], right_on=['node_index'], how='left').drop(columns=['node_index', 'node_type', 'node_name']).rename(columns={'node_id': 'tail_id', 'node_source': 'tail_source'})
```

```python
df.to_csv('link_result_rotate_200_primekg_dmd_indication.csv', index=False)
```
