```python
import pandas as pd
import random
```

```python
df = pd.read_csv('../../primekg/kg/nodes.csv')
df.head()
```

```python
with open('../../primekg/kg/entity2id.txt', 'w') as f:
    f.write(f'{len(df)}\n')
    for _, row in df.iterrows():
        f.write(f'{row["node_name"]}:{row["node_type"]}\t{row["node_index"]}\n')
```

```python
df = pd.read_csv('../../primekg/kg/edges.csv')
df.head()
```

```python
relations = df.drop_duplicates(subset=['relation'])['relation'].to_list()
relation2id = {relation: i for i, relation in enumerate(relations)}

with open('../../primekg/kg/relation2id.txt', 'w') as f:
    f.write(f'{len(relations)}\n')
    for k, v in relation2id.items():
        f.write(f'{k}\t{v}\n')
```

```python
df['rela_id'] = df['relation'].apply(lambda x: relation2id[x])
df = df[['x_index', 'y_index', 'rela_id']]
data = list(df.itertuples(index=False, name=None))
random.shuffle(data)
data[:10]
```

```python
with open('../../primekg/kg/all2id.txt', 'w') as f:
    f.write(f'{len(data)}\n')
    for x, y, r in data:
        f.write(f'{x} {y} {r}\n')
```

```python
train_size = int(len(data) * 0.9)
valid_size = int(len(data) * 0.05)
test_size = len(data) - train_size - valid_size

print('train size: ', train_size)
print('valid size: ', valid_size)
print('test size: ', test_size)

train_data = data[:train_size]
valid_data = data[train_size:train_size+valid_size]
test_data = data[train_size+valid_size:]
```

```python
with open('../../primekg/kg/train2id.txt', 'w') as f:
    f.write(f'{len(train_data)}\n')
    for x, y, r in train_data:
        f.write(f'{x} {y} {r}\n')
```

```python
with open('../../primekg/kg/valid2id.txt', 'w') as f:
    f.write(f'{len(valid_data)}\n')
    for x, y, r in valid_data:
        f.write(f'{x} {y} {r}\n')
```

```python
with open('../../primekg/kg/test2id.txt', 'w') as f:
    f.write(f'{len(test_data)}\n')
    for x, y, r in test_data:
        f.write(f'{x} {y} {r}\n')
```
